### Example 1: From Yellowbrick -documentation (classification)

```python
from yellowbrick.datasets import load_game
from yellowbrick.target import ClassBalance

# Load the classification dataset
X, y = load_game()

# Instantiate the visualizer
visualizer = ClassBalance(labels=["draw", "loss", "win"])

visualizer.fit(y)        # Fit the data to the visualizer
visualizer.show()        # Finalize and render the figure
```

This looks very familiar... just like seaborn bar chart + hue???

### Okay, this is pretty nice for seeing if our train/test -datasets are balanced

The test dataset is usually smaller, so it's more important to look if the distribution shape matches.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# load data
df = pd.read_csv("../lifestyle_sustainability_data.csv")

# handle these better with your own dataset, here we are simply dropping these
df = df.drop(['ParticipantID'], axis=1)

# Define features and target
X = df.drop("Rating", axis=1)
y = df['Rating']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# Instantiate the visualizer
visualizer = ClassBalance(labels=["1", "2", "3", "4", "5"])

visualizer.fit(y_train, y_test)        # Fit the data to the visualizer
visualizer.show()        # Finalize and render the figure
```

Rating 3 is somehow underrepresented when compared to training data.
