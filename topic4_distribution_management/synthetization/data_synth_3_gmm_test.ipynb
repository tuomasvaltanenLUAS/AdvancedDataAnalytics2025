```python
import pandas as pd
df = pd.read_csv("../houses.csv")
# handle these better with your own dataset, here we are simply dropping these
df = df.drop(['id', 'date', 'zipcode', 'lat', 'long'], axis=1)

import seaborn as sns
sns.histplot(data=df, x="price", kde=True)
```

```python
df
```

```python
# Install necessary libraries
# pip install sklearn

from sklearn.mixture import GaussianMixture
import pandas as pd
import numpy as np

# Initialize the GMM model with desired number of components
gmm = GaussianMixture(n_components=3)

# Fit the GMM model to the data
gmm.fit(df)

# Generate synthetic data from the GMM model
synthetic_data, _ = gmm.sample(n_samples=5000)

# Convert synthetic data back to DataFrame
synthetic_df = pd.DataFrame(synthetic_data, columns=df.columns)

print(synthetic_df)
```
